## Задание 1. Арифметические операции, ввод/вывод, форматируемые строки (0,5 балла)

**Задача:**

Напишите программу, которая:
1. Запрашивает у пользователя два целых числа.
2. Выполняет над этими числами 6 арифметических операций (+, -, *, /, %, //).
3. Выводит с помощью форматируемых строк результаты выполнения операций в следующем виде:<br>
`6 + 13 = 19`<br>
`6 - 13 = -7`<br>
`6 * 13 = 78` <br>
и т.д.<br>
Вещественные значения должны выводиться с точностью 3 знака после десятичной точки.

```python
# Место для решения задания 1
a=int(input("enter first number"))
b=int(input("enter second number"))
print(f"{a} + {b} = {a+b}")
print(f"{a} - {b} = {a-b}")
print(f"{a} * {b} = {a*b}")
print(f"{a} / {b} = {a/b:.3f}")
print(f"{a} % {b} = {a%b}")
print(f"{a} // {b} = {a//b}")
```

## Задание 2. Арифметические операции, условный оператор (1 балл)

**Задача:**

Васе нужно снять в банкомате определенную сумму. В банкомате есть купюры номиналом 50, 100, 500, 1000 и 5000 рублей. Банкомат выдает деньги таким образом, чтобы количество купюр было минимальным. Посчитайте, сколько купюр каждого достоинства будет выдано Васе. Предусмотреть ситуацию, когда выдача данной суммы невозможна (например, 210 рублей).
Вывести результаты в виде:<br>
`Исходная сумма - 11500 рублей`  
 `5000 рублей - 2 купюры`   
 `1000 рублей - 1 купюра`  
 `500 рублей - 1 купюра`  
 Если какие-то купюры не выдавались, то их писать не нужно.

```python
                                                  # Место для решения задания 2
a = int(input(""))
ca = a
r = [5000,1000,500,100,50]
result = [0,0,0,0,0]
print(f"Исходная сумма - {a}")
for i in range(0,5):
    if a==0:
        break
    if a//r[i]!=0:
        result[i] = a//r[i]
        a=a-r[i]*(a//r[i])
if a!=0:
    print(f"выдача суммы {ca} - невозможна")
else:
    for i  in range(0,5):
        if result[i] !=0:
            print(f"{r[i]} рублей - {result[i]} купюры")
```

## Задание 4. Цикл `for` и `while`(1 балл)

**Задача:**

С клавиатуры вводится N целых чисел. Если вводимое число имеет сумму цифр больше 10, то вывести это число и его сумму цифр. Посчитать, сколько таких чисел получилось.
**Не использовать функции и списки!**




```python
n = int(input("введите кол-во целых чисел"))
sum = 0
for i in range(0,n):
    a = int(input("введите число"))
    ca = a
    suma=0
    while a!=0:
        suma=suma+a%10
        a=a//10
    if suma>10:
        print(ca,suma)
        sum=sum+1
print(f"Всего таких числе получилось {sum}")
```
