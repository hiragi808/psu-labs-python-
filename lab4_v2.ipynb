## Задание 1. Генераторы и итераторы (1 балл)

**Темы:** итераторы и генераторы .


**Что нужно сделать:**
Напишите функцию-генератор fibonacci(n), которая генерирует первые n чисел Фибоначчи.
Затем создайте итератор evenIndexIterator, который принимает любую последовательность и возвращает только элементы, стоящие на чётных индексах.  
Проверьте работу программы на следующих примерах:   
- для Фибоначчи на последовательности `[0,1,2,5,8,15]`  
- для итератора на последовательностях `[10, 20, 30, 40, 50], ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i'], [1, "two", 3.0, [4, 5], (6, 7), {8: 9}, True, None]`      


```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
fib_examples = [0, 1, 2, 5, 8, 15]
for n in fib_examples:
    # Преобразуем генератор в список для наглядного вывода
    print(f"n = {n}: {list(fibonacci(n))}")
class EvenIndexIterator:
    def __init__(self, sequence):
        self.sequence = sequence
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.sequence):
            raise StopIteration
        result = self.sequence[self.index]
        self.index += 2  # Шаг через один элемент
        return result
print("\n=== Проверка итератора EvenIndexIterator ===")
iterator_examples = [
    [10, 20, 30, 40, 50],
    ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i'],
    [1, "two", 3.0, [4, 5], (6, 7), {8: 9}, True, None]
]

for seq in iterator_examples:
    print(f"Исходная последовательность: {seq}")
    print(f"Элементы на чётных индексах: {list(EvenIndexIterator(seq))}\n")
```

## Задание 2. Модуль NumPy (1 балл)

**Темы:** работа с одномерным массивом

**Что нужно сделать:**

Создайте массив из 20 случайных целых чисел от -100 до 100. Выполните следующие действия:
- Найдите минимальное, максимальное и среднее значение.  
- Получите массив из отрицательных элементов.
- Замените все значения, больше 10, на 0.
- Вычислите среднее аримфметическое элементов, находящихся на нечётных позициях.

```python
# Ваш код здесь
import numpy as np

# Создаем массив из 20 случайных чисел от -100 до 100
arr = np.random.randint(-100, 101, size=20)
print(f"Исходный массив:\n{arr}\n")

# 1. Поиск минимума, максимума и среднего
print("1. Статистика:")
print(f"   Минимум: {np.min(arr)}")
print(f"   Максимум: {np.max(arr)}")
print(f"   Среднее: {np.mean(arr)}")

# 2. Фильтрация отрицательных элементов
neg_arr = arr[arr < 0]
print(f"2. Отрицательные элементы:{neg_arr}")

# 3. Среднее арифметическое элементов на нечетных индексах (1, 3, 5...)
odd_pos = arr[1::2]
print(f"3. Элементы на нечетных позициях:{odd_pos}")
print(f"   Их среднее: {np.mean(odd_pos)}")

# 4. Замена значений больше 10 на 0
mod_arr = arr.copy()
mod_arr[mod_arr > 10] = 0
print(f"4. Замена элементов (>10 на 0):{mod_arr}")
```

## Задание 3. Модуль NumPy (1 балл)

**Темы:** работа с двумерным массивом

**Что нужно сделать:**
Создайте две матрицы A (3×4) и B (4×3), заполненные случайными числами от 0 до 9. 
Выполните:  
- Умножение матриц A * B (матричное произведение).  
- Поэлементное умножение A на транспонированную B.  
- Найдите среднее значение каждой строки матрицы A.


```python
import numpy as np

# Создаем матрицы случайных чисел от 0 до 9
A = np.random.randint(0, 10, size=(3, 4))
B = np.random.randint(0, 10, size=(4, 3))

print(f"Матрица A (3x4):\n{A}")
print(f"Матрица B (4x3):\n{B}")

# 1. Матричное произведение A * B (размерность результата: 3x3)
matrix_prod = np.dot(A, B)  #
print(f"1. Матричное произведение A * B:\n{matrix_prod}")

# 2. Поэлементное умножение A на транспонированную B (размерность обеих: 3x4)
B_T = B.T
element_prod = A * B_T
print(f"2. Транспонированная матрица B (3x4):\n{B_T}")
print(f"   Поэлементное умножение A * B.T:\n{element_prod}")

# 3. Среднее значение каждой строки матрицы A
# axis=1 означает вычисление вдоль строк
row_means = np.mean(A, axis=1)
print(f"3. Среднее значение каждой строки матрицы A:\n{row_means}")
```

## Задание 4. Работа с датой и временем (2 балла)

**Темы:** Модуль datetime

**Что нужно сделать:**
Программа просит пользователя ввести дату своего рождения в формате ДД.ММ.ГГГГ (например, 15.08.2003). Нужно определить:  
- Какой это был день недели (название на русском или английском).  
- Был ли этот год високосным.  
- Сколько пользователю полных лет на текущий момент (с учетом прошедших дней рождения).  

Пример ожидаемого вывода:

`Введите дату рождения (дд.мм.гггг): 15.08.2003`  
`Вы родились в пятницу.`  
`2003 год не был високосным.`  
`Сейчас вам 20 полных лет.`  

```python
import calendar
from datetime import datetime

# 1. Читаем дату из ввода
user_input = input("Введите дату рождения (дд.мм.гггг): ")
birth_date = datetime.strptime(user_input, "%d.%m.%Y")

# 2. Название дня недели на русском
days = ["понедельник", "вторник", "среду", "четверг", "пятницу", "субботу", "воскресенье"]
day_name = days[birth_date.weekday()]
print(f"Вы родились в {day_name}.")

# 3. Високосный год
year = birth_date.year
if calendar.isleap(year):
    print(f"{year} год был високосным.")
else:
    print(f"{year} год не был високосным.")

# 4. Считаем полные года
today = datetime.now()
age = today.year - birth_date.year

# Если в этом году дня рождения еще не было, вычитаем 1 год
if (today.month, today.day) < (birth_date.month, birth_date.day):
    age -= 1

print(f"Сейчас вам {age} полных лет.")
```
